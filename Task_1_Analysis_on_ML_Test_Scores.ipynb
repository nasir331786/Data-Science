```python
# Extract numeric score from Score column
df['Score_Numeric'] = df['Score'].str.split('/').str[0].astype(float)

# Batch-wise analysis
print("="*60)
print("BATCH-WISE ANALYSIS OF ML TEST SCORES")
print("="*60)

for batch in df['Batch'].unique():
    batch_data = df[df['Batch'] == batch]['Score_Numeric']
    print(f"\n{batch}:")
    print(f"  Number of students: {len(batch_data)}")
    print(f"  Mean score: {batch_data.mean():.2f}")
    print(f"  Median score: {batch_data.median():.2f}")
    print(f"  Standard deviation: {batch_data.std():.2f}")
    print(f"  Min score: {batch_data.min():.0f}")
    print(f"  Max score: {batch_data.max():.0f}")
    print(f"  Score distribution:")
    for score in sorted(batch_data.unique()):
        count = (batch_data == score).sum()
        percentage = (count / len(batch_data)) * 100
        print(f"    Score {score:.0f}: {count} students ({percentage:.1f}%)")

print("\n" + "="*60)
```

```python
# Create visualizations
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('ML Test Scores Analysis - Batch Comparison', fontsize=16, fontweight='bold')

# 1. Box plot comparison
ax1 = axes[0, 0]
df.boxplot(column='Score_Numeric', by='Batch', ax=ax1)
ax1.set_title('Score Distribution by Batch (Box Plot)')
ax1.set_xlabel('Batch')
ax1.set_ylabel('Score (out of 7)')

# 2. Histogram
ax2 = axes[0, 1]
for batch in sorted(df['Batch'].unique()):
    batch_scores = df[df['Batch'] == batch]['Score_Numeric']
    ax2.hist(batch_scores, alpha=0.5, label=batch, bins=7)
ax2.set_title('Score Distribution Histogram')
ax2.set_xlabel('Score')
ax2.set_ylabel('Frequency')
ax2.legend()
ax2.grid(alpha=0.3)

# 3. Mean and Median comparison
ax3 = axes[1, 0]
batch_stats = df.groupby('Batch')['Score_Numeric'].agg(['mean', 'median'])
batch_stats.plot(kind='bar', ax=ax3)
ax3.set_title('Mean and Median Scores by Batch')
ax3.set_xlabel('Batch')
ax3.set_ylabel('Score')
ax3.legend(['Mean', 'Median'])
ax3.grid(alpha=0.3)

# 4. Student count and overall performance
ax4 = axes[1, 1]
batch_count = df.groupby('Batch')['Score_Numeric'].count()
batch_mean = df.groupby('Batch')['Score_Numeric'].mean()
ax4.bar(range(len(batch_count)), batch_count, alpha=0.7, label='Student Count')
ax4_twin = ax4.twinx()
ax4_twin.plot(range(len(batch_mean)), batch_mean, 'ro-', linewidth=2, markersize=8, label='Mean Score')
ax4.set_xlabel('Batch')
ax4.set_ylabel('Number of Students')
ax4_twin.set_ylabel('Mean Score')
ax4.set_title('Student Count vs Mean Score')
ax4.set_xticks(range(len(batch_count)))
ax4.set_xticklabels(batch_count.index)
ax4.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\nVisualization complete!")
```
